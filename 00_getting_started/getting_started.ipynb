### Kernel state

The **kernel** is the Python process behind the notebook. Variables live in the kernel, not in the page.
If you run cells out of order, or restart the kernel, variables may be missing or stale.

Rule of thumb: when something behaves strangely, use **Runtime → Restart and run all**.

````python
x = 10
````

````python
x * 2   # works only after the cell above has been run
````

## 3. Checking the Python version and the main libraries

Colab comes with the standard data science stack already installed.

````python
import sys
import numpy as np
import pandas as pd
import matplotlib

print("Python    ", sys.version.split()[0])
print("NumPy     ", np.__version__)
print("pandas    ", pd.__version__)
print("Matplotlib", matplotlib.__version__)
````

## 4. Installing a package

If a library is missing, install it from a cell with `%pip install`. The `%` marks a *magic command* handled by Jupyter, not by Python.

```python
%pip install seaborn
```

Installed packages disappear when the Colab runtime is recycled, so keep the install cell at the top of your notebook.

## 5. Useful notebook commands

| Command | What it does |
|---|---|
| `len?` | show the help (docstring) of `len` |
| `%timeit expr` | time a short expression |
| `%who` | list the variables you have defined |
| `!ls` | run a shell command (Colab runs Linux) |

````python
%timeit sum(range(1000))
````

## 6. Loading data files

Course datasets are stored in the repository's `data/` folder. Notebooks read them directly from GitHub, so you do not need to upload anything:

````python
url = "https://raw.githubusercontent.com/alilouati/programming-for-data-science/main/data/courses.csv"
try:
    courses = pd.read_csv(url)
    display(courses)
except Exception as err:          # e.g. no internet connection
    print("Could not download the file:", err)
````

To use **your own file** in Colab, either

- click the folder icon on the left and drag the file in (it is deleted when the runtime stops), or
- mount Google Drive so files persist:

```python
from google.colab import drive

```

## 7. Working on your own computer (optional)

If you prefer to work offline:

1. Install Python 3.11 or newer from [python.org](https://www.python.org/downloads/), or install [Anaconda](https://www.anaconda.com/download) / [Miniforge](https://github.com/conda-forge/miniforge).
2. Clone the course repository and install the requirements:

```bash

```

## 8. A word on AI assistants

Colab, VS Code and other tools now include AI code assistants. They are useful for explaining errors and suggesting code, but:

- you are responsible for every line you submit, so **read and test** what they produce;
- they can be confidently wrong, especially about data: always check results against the data itself;
- follow the course policy on what is allowed in graded work.
